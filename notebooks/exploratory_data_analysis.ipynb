```python
%sql
use catalog merchai_data;
use schema gold;
```

```python
from pyspark.sql import functions as F
```

### Most Ordered Products

```python
top_products = (
    spark.table("gold_product_metrics")
         .orderBy(F.desc("total_orders"))
         .limit(10)
)
```

```python
display(top_products)
```

### Product with highest reorder rate

```python
top_reordered_products = (
    spark.table("gold_product_metrics")
         .filter(F.col("total_orders") > 100)
         .orderBy(F.desc("reorder_rate"))
         .limit(10)
)

```

```python
display(top_reordered_products)
```

Top department by order

```python
top_departments = (
    spark.table("gold_department_metrics")
         .orderBy(F.desc("total_orders"))
)
```

```python
display(top_departments.limit(10))
```

### Favourite departments

```python
favorite_departments = (
    spark.table("gold_customer_features")
         .groupBy("favorite_department")
         .count()
         .orderBy(F.desc("count"))
)
```

```python
display(favorite_departments)
```

### Favourite Aisles

```python
favorite_aisles = (
    spark.table("gold_customer_features")
         .groupBy("favorite_aisle")
         .count()
         .orderBy(F.desc("count"))
         .limit(15)
)
```

```python
display(favorite_aisles)
```

### Most active customers

```python
from pyspark.sql import functions as F

most_active_customers = (
    spark.table("gold_customer_features")
         .select(
             "user_id",
             "total_orders",
             "total_products",
             "avg_basket_size",
             "reorder_rate",
             "favorite_department",
             "favorite_aisle"
         )
         .orderBy(
             F.desc("total_orders"),
             F.desc("total_products")
         )
         .limit(20)
)
```

```python
display(most_active_customers)
```

### Basket size VS Reorder Rate

```python
basket_vs_reorder = (
    spark.table("gold_customer_features")
         .select(
             "avg_basket_size",
             "reorder_rate"
         )
)
```

```python
display(basket_vs_reorder)
```

### KPIs

```python
gold_customer = spark.table("gold_customer_features")
gold_product = spark.table("gold_product_metrics")
gold_department = spark.table("gold_department_metrics")

print(f"Total Customers      : {gold_customer.count():,}")
print(f"Total Products       : {gold_product.count():,}")
print(f"Total Departments    : {gold_department.count():,}")

gold_customer.select(
    F.avg("total_orders").alias("Avg Orders per Customer"),
    F.avg("avg_basket_size").alias("Avg Basket Size"),
    F.avg("reorder_rate").alias("Avg Reorder Rate")
).show(truncate=False)
```
